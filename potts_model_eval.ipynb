```python
import numpy as np
import torch
import matplotlib.pyplot as plt
from utils_potts import potts_2pt_corr_direction, visualize_potts, potts2d_ham, potts2d_swendsen_wang
from model import get_rope_vit_model, ExponentialMovingAverage
from utils_train import rnd
from utils import ess
from tqdm import tqdm

TABLEAU_COLORS = {
    'blue': '#1f77b4',
    'orange': '#ff7f0e',
    'green': '#2ca02c',
    'red': '#d62728',
    'purple': '#9467bd',
    'brown': '#8c564b',
    'pink': '#e377c2',
    'gray': '#7f7f7f',
    'olive': '#bcbd22',
    'cyan': '#17becf'
}
```

## Model Loading

```python
q = 3
L = 16
D = L**2
device = 'cuda:0'
J = 1
inv_temp = 'low'
beta = {'high': 0.5, 'crit': 1.005, 'low': 1.2}[inv_temp]
ckpt_dir = f"checkpoints/L_16_potts/potts_{inv_temp}.pth"

model = get_rope_vit_model(L, vocab_size=q+1, embed_dim=128, depth=4, num_heads=4, dtype='bfloat16', device=device)
ema = ExponentialMovingAverage(model.parameters(), decay=0.9999)
print('Model: num of params: {}, size: {:.2f} MB'.format(
    sum(p.numel() for p in model.parameters()),
    sum(p.numel() * p.element_size() for p in model.parameters()) / (1024 ** 2)))

checkpoint = torch.load(ckpt_dir, map_location=device, weights_only=False)

model.load_state_dict(checkpoint['model_state_dict'])
ema.load_state_dict(checkpoint['ema_state_dict'])
ema.store(model.parameters())
ema.copy_to(model.parameters())
model.eval();
```

## Generate Samples and Compute ESS

```python
log_rnd_list = []
samples_list = []
reward = lambda S: -beta * potts2d_ham(S, J=J, q=q)

## Generate in total 4096 samples
with torch.no_grad():
    for i in tqdm(range(4)):
        samples, log_rnd = rnd(model, reward, batch_size=1024, device=device)
        log_rnd_list.append(log_rnd)
        samples_list.append(samples)
        
log_rnd_list = torch.cat(log_rnd_list, dim=0)
samples_list = torch.cat(samples_list, dim=0)
print(f"Effective Sample Size = {ess(log_rnd_list):.4f}")
```

```python
# Ground Truth samples from Swendsen-Wang algorithm
swen_samples = potts2d_swendsen_wang(L=L, J=J, beta=beta, B=128, num_collect=32, burn_in=2 ** 10, collect_every=128, init=None)
```

## Visualization of Generated Samples

```python
fig1 = visualize_potts(samples_list[:25], q, 5, 5)
```

```python
fig2 = visualize_potts(swen_samples[:25], q, 5, 5)
```

## Visualization of 2pt Correlation

```python
mdns_corr_x = np.array([potts_2pt_corr_direction(samples_list, q=q, r_x=r, r_y=0, use_x=True, use_y=False).cpu().numpy() for r in range(-L//2,L//2)])
swen_corr_x = np.array([potts_2pt_corr_direction(swen_samples, q=q, r_x=r, r_y=0, use_x=True, use_y=False) for r in range(-L//2,L//2)])

print(mdns_corr_x)
print(swen_corr_x)

fig, ax = plt.subplots(nrows=1, ncols=1, figsize=(4, 4))

ax.plot(range(-L//2,L//2), mdns_corr_x, color=TABLEAU_COLORS['red'], linestyle='--', marker=">", label='MDNS')
ax.plot(range(-L//2,L//2), swen_corr_x, color=TABLEAU_COLORS['blue'], linestyle='-', marker="<", label='SW')

ax.set_xticks(np.arange(-L//2,L//2, 2))
ax.set_xlabel(r'Distance $r$', fontsize=14)
ax.set_ylabel(r'2-point Correlation', fontsize=14)
ax.set_title("Two-point Correlation of Potts Model", fontsize=14)
ax.legend(fontsize=12)

plt.subplots_adjust(wspace=0.1, hspace=0.1, bottom=0.2)
plt.tight_layout()
plt.show()
```
